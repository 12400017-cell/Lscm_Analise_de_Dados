1. Crie um vetor NumPy com as idades `[18, 21, 19, 22, 20]`.
2. Crie uma **matriz** 2×3 qualquer e imprima `shape` e o elemento `[1, 2]`.
3. Transforme as idades numa **Series** com índice `Ana, Bruno, Carla, Diego, Eva`.
4. Com `loc`, mostre a idade da Carla.
5. Filtre quem tem **21 anos ou mais**.

```python
import numpy as np
import pandas as pd

#print("numpy ", np.__version__)
#print("pandas ", pd.__version__)

#Criar um VEROT(1d) com as idades listadas-----------------------
numeros = np.array([18,21,19,22,20]) #FEITO
print("Idade: ", numeros)
print("Shape: ",numeros.shape) #(,5) --> pois tem 5 colunas e é unidimencional

#Crair matriz 2 linhas, 3 colunas(2x3) --> devo imprimir o shape + imprimir o elemento [1,2]------
num2 = np.array([
    [1,2,3],
    [4,5,6]
    ]
)
print("pessoas: ", num2.shape) #FEITO
print("Número da posição[1,2]: ", num2[1,2]) #FEITO


#Crie um series com index e valor e transforme as idades
idades = pd.Series(
    [18,21,19,22,20],
    index=["Ana","Bruno","Carla","Diego","Eva"],
    name="Idades"
) #FEITO

#Mostrar idade de carla com iloc e loc
print("Carla(Iloc): ", idades.iloc[2])
print("Carla(loc): ", idades.loc["Carla"])

#Mostrar pessoas com idade >= 21

```
