```python
#Criando vetor numpy
#importando andas e a biblioteca numPy
import numpy as np
import pandas as pd

#verificando
print("pandas: ", pd.__version__)
print("numpy: ", np.__version__)


#1- crian um vetor NumPy com as idades 18,21,19,22,20
vetorIdades = np.array([18,21,19,22,20])
print("Tipo de variavel: ",type(vetorIdades))
print("Numero dimenções(ndim): ", vetorIdades.ndim)
print("Média: ", vetorIdades.mean())
```

```python
# 2- criar uma MATRIZ 2x3 que printe o shape de [1,2] (1 linha, 2 colunas)
# Uma matriz possui 2 dimenções: linhas, culunas --> shape
matrizNumeros = np.array(
    [
        [10,20,30],
        [20,30,40]
    ]
)
print("Shape-Geral: ", matrizNumeros.shape) #1 linha, 2 elementos
print("Shepe-Elemento[1,2]:", matrizNumeros[1,2])
```

```python
# 3- Transforme as idades em uma serie da indices Ana, Bruno, Carla, Diego, Eva

# pd.Series é um vetor(tipo NumPy) com um índice
# values --> os Dados
# index --> nome de cada posição 
# name --> nome da série(opcional)
#se não dar um indice, o pandar atribuira 0,1,2,3...
matrizNomeIdade= pd.Series(
    [10,20,30,40,50], #Atribuindo valores
    index = ["Ana","Bruno","Carla","Diego","Eva"], #atribuindo index
    name = "Nome nota"
)
print("Questão 3 ---------------------------")
print(matrizNomeIdade)

# 4 - Com loc, mostre a idade da Carla
# loc --> rotulo do indice
# iloc --> posição
print("\nQuestão 4 ---------------------------")
print("Carla(loc):",matrizNomeIdade.loc["Carla"])

#printe quem tem +de 21 anos
print("\nQuestão 5 ---------------------------")

```
