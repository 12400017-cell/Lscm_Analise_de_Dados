```python
# No terminal, uma vez: python -m pip install openpyxl pandas
import pandas as pd
import openpyxl

path = r"C:\Users\Leonardo Melo\Desktop\Pytom(Jenaina)\Repositorio-GitHub\Lscm_Analise_de_Dados\Atv3_TODO\Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
#Organizando o Exel--------------------------------------------------------------
tbBruta = pd.read_excel(path, header=None)
tbBruta.columns = [ #renomendo colunas
    "Estado",
    "Total TODOS",
    "H-Geral",
    "M-Geral",
    "Total-Exatas",
    "H-Exatas",
    "M-Exatas",
    "Total Duração-Serviço",
    "H-Duração",
    "M-Duração"
]
tbBruta = tbBruta.iloc[3:36]
#Para a parte final da atividade, é nescessário que eu use o desbribe, mas ele precisa que os valores da tabela toda sejam usaveis como números
#por tanto, já de começo, devo converter a tabela toda(menos estados) pra numero
#Importante: para modificar os valores retidos nas colunas uso o APPLY()
#ou seja, para mudar tudo, devo usar o .iloc para acessar todas as colunas(menos estado)
colPresente = ["Total TODOS", "H-Geral", "M-Geral","Total-Exatas", "H-Exatas", "M-Exatas","Total Duração-Serviço", "H-Duração", "M-Duração"]
tbBruta[colPresente] = tbBruta[colPresente].apply(pd.to_numeric)
```

```python
#Questão 1--------------------------------------------------------------
#Homens e mulheres por região com curso superior
q1 = tbBruta
q1 = q1.iloc[4:9] #Limitando as regiões
#tbBruta.loc[linha,coluna]
q1 = q1.loc[:,'Estado':'M-Geral'] #ou tbBruta[['Estado','Total TODOS','H-Geral','M-Geral']]
```

```python
#Questão 2--------------------------------------------------------------
#homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
q2 = tbBruta
q2 = q2.iloc[4:9]
q2 = q2[['Estado','Total-Exatas','H-Exatas','M-Exatas']]
```

```python
#Questão 3--------------------------------------------------------------
#Homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar 
q3 = tbBruta
q3 = q3.iloc[4:9]   
q3 = q3[['Estado','Total Duração-Serviço','H-Duração','M-Duração']]
q3
```
